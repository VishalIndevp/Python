# Python supports 3 categories of data types: 
##### • Basic Types - (integer,float,complex,boolean and string)
##### • Container Types - (List, Tuples, Sets and Dictionary)
##### • User-defined Types - Class

```python
# Interger 
print(5)
print(1e308) # last limit
```

```python
#float 
print(5.5)
print(1.7e308)
```

```python
# bolean 
print(True) 
print(False)
```

```python
#Complex 
print(4+5j)
```

```python
# String 
print('Hello')
print("Hello")
print("""Hello""")
```

## Container Types 

```python
#list 
print([1,2,3,4,5,6])
```

```python
#tuple 
print((1,2,3,4,5))
```

```python
#Sets 
print({1,2,3,4,5})
```

```python
# Dict
print({'Name':"Vishal Singh","Age":21})
```
