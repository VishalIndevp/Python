# Operators in Python
#### Arithmetic: +, -, *, /, %, **, //
##### Comparison: ==, !=, >, <, >=, <=
##### Logical: and, or, not
##### Bitwise: &, |, ^, ~, <<, >>
##### Assignment: =, +=, -=, *=, /=, %=, **=, //=
##### Identity: is, is not
##### Membership: in, not in

# Bitwise Operators

```python
print(x & y) # AND
```

```python
print(x | y) # OR
```

```python
print(2 ^ 3) # XOR
```

```python
print(x >> 2) # Right Shift
```

```python
print(y << 3) # Left Shift
```

```python
print(~ x) # NOT
```

# Assignment Operators 

```python
a = 3
print(a)
```

```python
a += 3   # a = a + 3
print(a)
```

```python
a -= 3   # a = a - 3
print(a)
```

```python
a *= 3   # a = a * 3
print(a)
```

```python
a /= 3   # a = a / 3
print(a)
```
