# While Loop 

```python
num= int(input("Enter the Number"))
i=1

while i<11:
    print(num *i)
    i+=1
```

```python
# Better Way
num= int(input("Enter the Number"))
i=1

while i<11:
    print(num,'*',i,num *i)
    i+=1
```
